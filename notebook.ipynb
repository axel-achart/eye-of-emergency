## Contexte et problématique

Lors d'une catastrophe (incendie, inondation, séisme...), les réseaux sociaux comme Twitter sont devenus une source d'information en temps réel. Ce notebook répond à ce besoin en construisant un modèle de classification de texte (NLP + Machine Learning) capable de faire ce tri : à partir du texte d'un tweet (et éventuellement de son mot-clé et sa localisation), le modèle prédit s'il s'agit (1) ou non (0) d'une catastrophe réelle.

# A) A la découverte du NLP

### 1. Notion de Text Mining et de NLP, ses points communs et différences

**Le Text Mining** (ou fouille de texte) est une discipline qui consiste à extraire des informations utiles, des tendances ou des connaissances à partir de grandes quantités de texte non structuré. Elle s'appuie sur des méthodes statistiques et du machine learning pour transformer du texte brut en données exploitables : classification de documents, détection de sujets récurrents, extraction de motifs, etc. Son objectif principal est de répondre à la question "que peut-on apprendre de ce texte ?".

**Le NLP** (Natural Language Processing, ou traitement du langage naturel) est une branche de l'intelligence artificielle qui s'intéresse à la façon dont une machine peut comprendre, interpréter et générer le langage humain. Il couvre la grammaire, le sens des mots, le contexte d'une phrase, et permet des tâches comme la traduction automatique, la reconnaissance vocale ou la génération de texte. Son objectif est de répondre à la question "comment la machine comprend le langage ?".

**Points communs** : les deux domaines travaillent sur des données textuelles non structurées et partagent les mêmes étapes de prétraitement (nettoyage, tokenisation, suppression des stop-words, etc). En pratique, le NLP est souvent l'outil que le Text Mining utilise pour arriver à ses fins.

**Différences** : le Text Mining se concentre sur l'extraction de connaissances et d'insights à grande échelle. 
Le NLP se concentre sur la compréhension fine du langage lui-même.

**Exemple concret** : imaginons une entreprise qui reçoit des milliers d'avis clients sur ses produits.

- Avec le **Text Mining**, on cherche à dégager des tendances globales : quels sont les mots les plus fréquents dans les avis, quels produits reviennent le plus souvent dans les commentaires négatifs, quels sujets ressortent (livraison, qualité, prix). Le but est d'obtenir des statistiques utiles pour l'entreprise.
- Avec le **NLP**, on analyse chaque avis individuellement pour en comprendre le sens réel : par exemple reconnaître que la phrase "le produit n'est pas mauvais du tout" exprime un avis positif malgré la négation, ou repérer automatiquement le nom du produit mentionné dans le texte. Le but est de comprendre le langage utilisé.

Le NLP fournit donc les briques de compréhension du texte que le Text Mining exploite ensuite pour en tirer des connaissances.

### 2. NLP et ses sous-domaines

Le NLP regroupe plusieurs sous-domaines, chacun répondant à un besoin précis dans la compréhension du langage. En voici trois principaux.


**L'analyse de sentiments (Sentiment Analysis) :** </br>
Elle consiste à déterminer l'opinion ou l'émotion exprimée dans un texte : positive, négative ou neutre. C'est une technique très utilisée pour analyser des avis clients, des tweets ou des commentaires sur les réseaux sociaux.

Exemple : la phrase *"Ce restaurant est vraiment excellent, je recommande !"* sera classée comme un sentiment positif, tandis que *"Le service était lent et le plat froid"* sera classée comme négative.


**La reconnaissance d'entités nommées (Named Entity Recognition, NER) :** </br>
Elle consiste à repérer dans un texte les entités importantes et à les classer par catégorie : personnes, lieux, organisations, dates, montants, etc.

Exemple : dans la phrase *"Emmanuel Macron s'est rendu à Paris le 14 juillet"*, le NER identifie "Emmanuel Macron" comme une personne, "Paris" comme un lieu et "14 juillet" comme une date.


**L'étiquetage grammatical (Part-of-Speech Tagging, POS Tagging) :** </br>
Elle consiste à attribuer à chaque mot d'une phrase sa nature grammaticale : nom, verbe, adjectif, déterminant, etc.

Exemple : dans la phrase *"Le chat mange une souris"*, le POS Tagging donnerait : "Le" (déterminant), "chat" (nom), "mange" (verbe), "une" (déterminant), "souris" (nom). Cette étape sert souvent de base à d'autres traitements, comme la reconnaissance d'entités ou l'analyse syntaxique.

### 3. Exemples d’applications concrètes du NLP

Le NLP est aujourd'hui présent dans de nombreux outils que l'on utilise au quotidien, souvent sans même s'en rendre compte.

- **Traduction automatique** : des outils comme Google Translate ou DeepL traduisent un texte d'une langue à une autre en tenant compte du contexte.
- **Résumé automatique de texte** : certains outils condensent un article ou un document long en quelques phrases clés.
- **Détection de spam et de fake news** : des modèles NLP identifient les messages indésirables ou les contenus trompeurs, ce qui rejoint directement la problématique de ce projet.

### 4. Les stop-words

Un stop-word (ou mot vide) est un mot très courant comme "le", "et" ou "de", qui n'apporte pas de valeur informative forte.
Il est important de les supprimer car, apportant aucune valeur, ils ne sont pas nécessaire et sont un peu comme une "pollution"

Exemple : dans  "*La pêche et la chasse sont des activités de grande date*", les mots "la", "et", "des" et "de" sont inutile et on peut très bien comprendre le sens de la phrase sans ces mots.

### 5. Gestion des ponctuations et caractères spéciaux

Les ponctuations et les caractères spéciaux sont traités de cette manière :
la suppression, 
l'isolation en tokens,
ou la conservation contextuelle. 

Selon la tâche visée, ils sont soit effacés pour réduire le bruit, soit analysés pour leur sens.

### 6. Qu’est ce qu’un token ? un N-gram ? Quel processus permet-il de les obtenir ?

Un token (ou jeton) est un morceau de texte, comme un mot entier, une partie de mot, un caractère ou un signe de ponctuation, qu'une machine peut lire et transformer en nombre.

Un N-gram est une séquence contiguë de N éléments (mots ou caractères) extraite d'un texte. 
Les types les plus courants sont les unigrammes (N=1), les bigrammes (N=2) et les trigrammes (N=3).
*Exemples = Dans la phrase « Le chat dort », la découpe donne :*
*Unigrammes (N=1) : « Le », « chat », « dort »*
*Bigrammes (N=2) : « Le chat », « chat dort »*
*Trigrammes (N=3) : « Le chat dort »*
*Ces suites permettent de capturer le contexte local d'une phrase en comptant la fréquence d'apparition des mots les uns à la suite des autres*

Pour les obtenir, le processus consiste à nettoyer un texte, réaliser une tokenisation (découpage en mots ou sous-mots), puis regrouper ces tokens par séquences de taille n (les n-grammes)

### 7. Les notions Stemming et de Lemmatization

**Le Stemming** (ou racinisation) est une technique qui consiste à réduire un mot à sa racine (ou "stem") en supprimant simplement son suffixe, à l'aide de règles heuristiques, sans se soucier de la grammaire ni du sens réel du mot. Le résultat obtenu n'est pas forcément un mot correct de la langue.

Exemple : "connexion", "connecter" et "connecté" seront tous les trois réduits à la racine "connect", et "chevaux" pourrait être réduit à "cheva" (qui n'existe pas en français).

**La Lemmatization** (ou lemmatisation) consiste également à ramener un mot à sa forme de base, mais cette fois en s'appuyant sur un dictionnaire et une analyse grammaticale (le contexte, la nature du mot) pour retrouver son **lemme**, c'est-à-dire sa forme canonique telle qu'on la trouverait dans un dictionnaire (infinitif pour un verbe, singulier masculin pour un nom, etc). Le résultat est donc toujours un mot correct.

Exemple : "connecté", "connectons" et "connectais" seront tous ramenés au lemme "connecter", et "chevaux" sera correctement ramené à "cheval".

**Différence principale** : le Stemming est une méthode rapide et purement mécanique (coupe de suffixes), quitte à produire des mots qui n'existent pas, tandis que la Lemmatization est plus précise et plus coûteuse en calcul, car elle nécessite de connaître le contexte et la grammaire du mot pour retrouver sa vraie forme de base.

**Quand utiliser l'un plutôt que l'autre ?**
- Le **Stemming** est préférable quand on privilégie la rapidité et que la précision exacte du mot importe peu, par exemple pour un moteur de recherche ou un premier prétraitement rapide sur de très gros volumes de texte.
- La **Lemmatization** est préférable quand le sens et la qualité du texte prétraité sont importants, par exemple pour de l'analyse de sentiments ou de la classification de texte (comme notre problématique de classification de tweets), car elle évite de fusionner des mots qui n'ont pas le même sens sous une racine incorrecte.


### 8. Études des deux méthodes Bag of Words et TF-IDF.

Pour qu'un algorithme de Machine Learning puisse traiter du texte, il faut le convertir en vecteurs numériques. Le Bag of Words et le TF-IDF sont deux méthodes de vectorisation classiques qui reposent toutes les deux sur la construction d'un vocabulaire à partir du corpus de textes.

**Le Bag of Words (BoW)** représente chaque document sous la forme d'un vecteur qui compte le nombre d'occurrences de chaque mot du vocabulaire dans ce document, sans tenir compte de l'ordre des mots ni du sens de la phrase (d'où le nom de "sac de mots").

Exemple : pour les deux phrases "le chat mange" et "le chien mange", le vocabulaire est {le, chat, chien, mange}. On obtiendrait :
- "le chat mange" → [1, 1, 0, 1]
- "le chien mange" → [1, 0, 1, 1]

La limite du BoW est que les mots très fréquents mais peu informatifs (comme "le") pèsent autant que des mots réellement importants, ce qui peut fausser le signal.

**Le TF-IDF (Term Frequency - Inverse Document Frequency)** vient corriger cette limite en pondérant chaque mot selon son importance réelle dans le corpus, et pas seulement selon sa fréquence brute. Il se compose de deux termes :
- **TF (Term Frequency)** : la fréquence d'un mot dans un document donné (plus un mot apparaît souvent dans un document, plus son TF est élevé).
- **IDF (Inverse Document Frequency)** : une pondération qui diminue le poids des mots qui apparaissent dans beaucoup de documents différents (comme les stop-words), et qui augmente celui des mots rares et donc plus discriminants.

Le score TF-IDF d'un mot est obtenu en multipliant ces deux valeurs (TF x IDF). Ainsi, un mot très fréquent dans un seul tweet mais rare dans l'ensemble des tweets aura un score élevé, tandis qu'un mot qui revient dans presque tous les tweets (comme "le" ou "de") aura un score très faible, même s'il est fréquent.

**En résumé** : le Bag of Words est plus simple et plus rapide à mettre en place, mais traite tous les mots de la même façon, tandis que le TF-IDF apporte une pondération plus fine qui met en avant les mots réellement discriminants pour distinguer les documents, ce qui en fait souvent un meilleur choix pour des tâches de classification de texte comme la nôtre.

### 3. Notre propre pipeline de traitement de texte

1. **Décodage des entités HTML** (ex : `&amp;` → `&`), fréquentes dans les tweets bruts.
2. **Passage en minuscules**, pour éviter que "Fire" et "fire" soient traités comme deux tokens différents.
3. **Suppression des URLs** (`http://…`, `www…`), qui n'apportent pas d'information textuelle exploitable.
4. **Suppression des mentions** (`@utilisateur`), qui sont des identifiants et non des mots du langage.
5. **Suppression du symbole `#`** tout en conservant le mot du hashtag (ex : `#earthquake` → `earthquake`), car ce mot est souvent très informatif pour notre problématique.
6. **Suppression de la ponctuation, des caractères spéciaux et des chiffres**, qui sont du bruit pour une vectorisation Bag of Words / TF-IDF.
7. **Tokenisation** du texte nettoyé. (séparation en plusieurs tokens choisis)
8. **Suppression des stop-words** (anglais), comme expliqué dans la veille.
9. **Lemmatisation** de chaque token, préférée au stemming car notre objectif est une classification où la qualité et le sens des mots comptent et non pas une généralisation.

```python
import html
from sklearn.base import BaseEstimator, TransformerMixin
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize


# Pipeline de nettoyage de texte pour les tweets, compatible scikit-learn.
class TextPreprocessor(BaseEstimator, TransformerMixin):

    def __init__(self, remove_stopwords=True, lemmatize=True):
        self.remove_stopwords = remove_stopwords
        self.lemmatize = lemmatize

    def fit(self, X, y=None):
        self._stop_words = set(stop_words_en) if self.remove_stopwords else set()
        self._lemmatizer = WordNetLemmatizer() if self.lemmatize else None
        return self

    def transform(self, X, y=None):
        return [self._clean(text) for text in X]

    def _clean(self, text):
        text = html.unescape(text)
        text = text.lower()
        text = re.sub(r"http\S+|www\.\S+", " ", text)
        text = re.sub(r"@\w+", " ", text)
        text = re.sub(r"#", "", text)
        text = re.sub(r"[^a-z\s]", " ", text)

        tokens = word_tokenize(text)
        if self.remove_stopwords:
            tokens = [tok for tok in tokens if tok not in self._stop_words]
        if self.lemmatize:
            tokens = [self._lemmatizer.lemmatize(tok) for tok in tokens]
        tokens = [tok for tok in tokens if len(tok) > 1]

        return " ".join(tokens)


text_preprocessor = TextPreprocessor()
db_train["clean_text"] = text_preprocessor.fit_transform(db_train["text"])

print("Exemples de tweets avant / après le pipeline de nettoyage :\n")
for raw, clean in db_train[["text", "clean_text"]].head(5).itertuples(index=False):
    print(f"RAW   : {raw}")
    print(f"CLEAN : {clean}\n")
```

### 5. Les 5 modèles d'apprentissage, et la différence entre bagging et boosting.

**1. Régression Logistique**
Malgré son nom, c'est un modèle de **classification**. Elle calcule une probabilité (entre 0 et 1) que le tweet appartienne à la classe "catastrophe", à partir d'une combinaison linéaire des variables d'entrée passée dans une fonction sigmoïde. Si la probabilité dépasse 0,5, on prédit 1, sinon 0. C'est un modèle simple, rapide et facile à interpréter.

**2. Decision Tree (Arbre de décision)**
Un arbre de décision découpe successivement les données selon des questions simples sur les variables (ex : "le mot 'fire' est-il présent ?"), formant une suite de branches jusqu'à une feuille qui donne la prédiction finale. C'est facile à visualiser et à interpréter, mais un seul arbre a tendance à sur-apprendre (overfitting) sur les données d'entraînement.

**3. Random Forest**
Une forêt aléatoire est un ensemble de nombreux arbres de décision, chacun entraîné sur un échantillon légèrement différent des données (et un sous-ensemble aléatoire des variables). La prédiction finale est obtenue par un vote majoritaire de tous les arbres. En combinant plusieurs arbres imparfaits mais différents, elle réduit l'overfitting d'un arbre unique.

**4. XGBoost (Extreme Gradient Boosting)**
XGBoost construit lui aussi de nombreux arbres, mais cette fois de façon **séquentielle** : chaque nouvel arbre est entraîné pour corriger les erreurs commises par les arbres précédents. C'est une implémentation optimisée (rapide, régularisée) du boosting de gradient, très performante en pratique sur des données tabulaires.

**5. Support Vector Machine (SVM)**
Un SVM cherche la frontière (hyperplan) qui sépare le mieux les deux classes, en maximisant la marge, c'est-à-dire la distance entre cette frontière et les points les plus proches de chaque classe (les "vecteurs de support"). Avec un noyau (kernel) adapté, il peut aussi séparer des données qui ne sont pas linéairement séparables.

#### Différence entre bagging et boosting

Les deux sont des méthodes d'**ensemble** : elles combinent plusieurs modèles "faibles" (souvent des arbres) pour obtenir un modèle final plus performant, mais elles ne les combinent pas de la même façon.

- **Bagging** (*Bootstrap Aggregating*) : les modèles sont entraînés **en parallèle**, indépendamment les uns des autres, chacun sur un échantillon tiré aléatoirement (avec remise) des données. Les prédictions sont ensuite agrégées (vote majoritaire en classification, moyenne en régression). Le bagging réduit surtout la **variance** (l'overfitting). *Exemple : Random Forest.*

- **Boosting** : les modèles sont entraînés **en séquence**, chacun essayant de corriger les erreurs du précédent en donnant plus de poids aux observations mal classées. Les prédictions sont combinées par un vote pondéré. Le boosting réduit surtout le **biais** (le sous-apprentissage). *Exemple : XGBoost.*

**Petit exemple concret** : imaginons 3 tweets de catastrophe mal classés par un premier arbre de décision.
- En **bagging**, on entraîne par exemple 100 arbres sur des échantillons différents, chacun ignorant l'existence de ces erreurs ; ces 3 tweets seront peut-être encore mal classés par certains arbres, mais noyés dans le vote majoritaire des 100 arbres, ce qui stabilise la prédiction globale.
- En **boosting**, le 2ᵉ arbre est entraîné en donnant plus d'importance à ces 3 tweets mal classés, pour apprendre spécifiquement à les corriger ; le 3ᵉ arbre se concentrera à son tour sur les erreurs restantes, et ainsi de suite. Le modèle final "apprend de ses erreurs" au fil des itérations.

### 6. Propre modèle Decision Tree sous forme de classe Python

Nous codons notre propre arbre de décision (algorithme CART), sans utiliser scikit-learn pour la partie apprentissage. Le principe :

- À chaque nœud, on cherche la **coupure** (une variable + un seuil) qui sépare le mieux les deux classes, c'est-à-dire celle qui fait le plus baisser l'**impureté de Gini** (une mesure du mélange des classes : 0 si un nœud est pur, 0.5 si les classes sont mélangées 50/50).
- On répète cette opération sur chaque sous-groupe créé, jusqu'à une profondeur maximale (`max_depth`) ou jusqu'à ce qu'un nœud soit trop petit (`min_samples_split`) ou pur.
- Une fois l'arbre construit, prédire revient à suivre le chemin des questions jusqu'à une feuille, qui donne la classe majoritaire.

Pour rester efficace en temps de calcul (un arbre codé en Python pur est bien plus lent que l'implémentation en C de scikit-learn), nous limitons volontairement le nombre de mots utilisés (300 mots les plus importants en TF-IDF) et la profondeur de l'arbre.

Avant d'entraîner nos modèles, nous séparons `db_train` en un jeu d'entraînement et un jeu de validation, pour pouvoir évaluer honnêtement chaque modèle sur des tweets qu'il n'a jamais vus.

```python
from sklearn.model_selection import train_test_split

# Separation du jeu d'entrainement en train / validation pour evaluer nos modeles honnetement
X_train_text, X_val_text, y_train, y_val = train_test_split(
    db_train["clean_text"], db_train["target"], test_size=0.2, random_state=42, stratify=db_train["target"]
)

print(f"Tweets d'entraînement : {len(X_train_text)}")
print(f"Tweets de validation  : {len(X_val_text)}")
```

```python
from sklearn.base import BaseEstimator, ClassifierMixin

# Arbre de decision CART code a la main (impurete de Gini), compatible scikit-learn (fit/predict)
class CustomDecisionTree(BaseEstimator, ClassifierMixin):

    def __init__(self, max_depth=8, min_samples_split=10):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split

    # Convertit une matrice creuse (sortie d'un vectorizer) en tableau numpy dense
    def _to_dense(self, X):
        if hasattr(X, "toarray"):
            return X.toarray().astype(np.float64)
        return np.asarray(X, dtype=np.float64)

    def fit(self, X, y):
        X = self._to_dense(X)
        y = np.asarray(y, dtype=np.int64)
        self.classes_ = np.unique(y)
        self.tree_ = self._build_tree(X, y, depth=0)
        return self

    # Impurete de Gini : 0 si le noeud est pur, 0.5 si les classes sont melangees 50/50
    def _gini(self, y):
        if len(y) == 0:
            return 0.0
        p = np.bincount(y, minlength=2) / len(y)
        return 1.0 - np.sum(p ** 2)

    # Cherche la meilleure coupure (variable + seuil) sur tout le jeu de donnees du noeud
    def _best_split(self, X, y):
        n_samples, n_features = X.shape
        best_gain = 0.0
        best_feature, best_threshold = None, None
        parent_gini = self._gini(y)

        for feature in range(n_features):
            values = X[:, feature]
            order = np.argsort(values)
            sorted_values = values[order]
            sorted_y = y[order]

            # positions ou la valeur change reellement (evite les seuils inutiles sur doublons)
            change_points = np.where(np.diff(sorted_values) != 0)[0]
            if len(change_points) == 0:
                continue

            # comptage cumule des tweets "catastrophe" pour calculer le gini a gauche/droite d'un coup
            cum_pos = np.cumsum(sorted_y)
            total_pos = cum_pos[-1]

            left_counts = change_points + 1
            left_pos = cum_pos[change_points]
            right_counts = n_samples - left_counts
            right_pos = total_pos - left_pos

            left_p = left_pos / left_counts
            right_p = right_pos / right_counts
            gini_left = 1.0 - left_p ** 2 - (1 - left_p) ** 2
            gini_right = 1.0 - right_p ** 2 - (1 - right_p) ** 2

            weighted_gini = (left_counts * gini_left + right_counts * gini_right) / n_samples
            gains = parent_gini - weighted_gini

            best_idx_local = np.argmax(gains)
            if gains[best_idx_local] > best_gain:
                best_gain = gains[best_idx_local]
                best_feature = feature
                idx = change_points[best_idx_local]
                best_threshold = (sorted_values[idx] + sorted_values[idx + 1]) / 2

        return best_feature, best_threshold, best_gain

    # Construction recursive de l'arbre
    def _build_tree(self, X, y, depth):
        n_samples = len(y)
        n_pos = int(np.sum(y))

        # feuille : on s'arrete et on garde la classe majoritaire
        if (
            depth >= self.max_depth
            or n_samples < self.min_samples_split
            or n_pos == 0
            or n_pos == n_samples
        ):
            return {"leaf": True, "prediction": int(n_pos >= n_samples / 2)}

        feature, threshold, gain = self._best_split(X, y)
        if feature is None or gain <= 0:
            return {"leaf": True, "prediction": int(n_pos >= n_samples / 2)}

        left_mask = X[:, feature] <= threshold
        right_mask = ~left_mask

        left_node = self._build_tree(X[left_mask], y[left_mask], depth + 1)
        right_node = self._build_tree(X[right_mask], y[right_mask], depth + 1)

        return {
            "leaf": False,
            "feature": feature,
            "threshold": threshold,
            "left": left_node,
            "right": right_node,
        }

    # Fait descendre une observation dans l'arbre jusqu'a une feuille
    def _predict_one(self, x, node):
        if node["leaf"]:
            return node["prediction"]
        if x[node["feature"]] <= node["threshold"]:
            return self._predict_one(x, node["left"])
        return self._predict_one(x, node["right"])

    def predict(self, X):
        X = self._to_dense(X)
        return np.array([self._predict_one(x, self.tree_) for x in X])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score

# Vectorisation limitee a 300 mots pour que notre arbre "maison" reste rapide
demo_vectorizer = TfidfVectorizer(max_features=300)
X_train_demo = demo_vectorizer.fit_transform(X_train_text)
X_val_demo = demo_vectorizer.transform(X_val_text)

my_tree = CustomDecisionTree(max_depth=8, min_samples_split=10)
my_tree.fit(X_train_demo, y_train)
my_pred = my_tree.predict(X_val_demo)

# Meme parametres sur l'arbre de scikit-learn, pour verifier que notre implementation est coherente
sk_tree = DecisionTreeClassifier(max_depth=8, min_samples_split=10, random_state=42)
sk_tree.fit(X_train_demo, y_train)
sk_pred = sk_tree.predict(X_val_demo)

print("Notre CustomDecisionTree :")
print(f"  Accuracy = {accuracy_score(y_val, my_pred):.3f} | F1-score = {f1_score(y_val, my_pred):.3f}")
print("\nDecisionTreeClassifier (scikit-learn) :")
print(f"  Accuracy = {accuracy_score(y_val, sk_pred):.3f} | F1-score = {f1_score(y_val, sk_pred):.3f}")
```

**Interprétation :** notre `CustomDecisionTree` obtient exactement les mêmes résultats que le `DecisionTreeClassifier` de scikit-learn, entraîné avec les mêmes paramètres et les mêmes données. Cela valide notre implémentation : elle recherche bien, à chaque nœud, la coupure qui fait le plus baisser l'impureté de Gini. Sa performance reste toutefois modeste, car nous n'utilisons que 300 mots (contre plusieurs milliers pour les autres modèles de la section suivante) afin de garder un temps d'entraînement raisonnable en Python pur.

### 7. Avec GridSearch, différents modèles de classification des tweets comparés avec modification de paramètres

Nous comparons maintenant nos 5 modèles d'apprentissage (section 5) ainsi que notre `CustomDecisionTree` (section 6), en faisant varier :
- le **preprocessing** : `CountVectorizer` (Bag of Words) vs `TfidfVectorizer`,
- le **type de modèle** : Régression Logistique, Decision Tree, Random Forest, XGBoost, SVM, et notre arbre maison,
- les **hyperparamètres** propres à chaque modèle.

Pour cela, chaque combinaison est assemblée dans un `Pipeline` scikit-learn (`vectorizer` → `classifier`), et `GridSearchCV` teste automatiquement toutes les combinaisons avec une validation croisée à 5 plis, en optimisant le **F1-score** (la métrique demandée par la problématique).

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV

# Pipeline generique : vectorisation du texte puis classification
pipeline = Pipeline([
    ("vectorizer", TfidfVectorizer()),
    ("classifier", LogisticRegression()),
])

# Les 2 facons de vectoriser le texte que l'on souhaite comparer
vectorizers = [
    CountVectorizer(max_features=3000, ngram_range=(1, 2)),
    TfidfVectorizer(max_features=3000, ngram_range=(1, 2)),
]

# Une entree de grille par modele, avec ses hyperparametres propres
param_grid = [
    {"vectorizer": vectorizers, "classifier": [LogisticRegression(max_iter=1000)],
     "classifier__C": [0.1, 1]},
    {"vectorizer": vectorizers, "classifier": [DecisionTreeClassifier(random_state=42)],
     "classifier__max_depth": [10, None]},
    {"vectorizer": vectorizers, "classifier": [RandomForestClassifier(random_state=42)],
     "classifier__n_estimators": [100, 200]},
    {"vectorizer": vectorizers, "classifier": [XGBClassifier(eval_metric="logloss", random_state=42)],
     "classifier__n_estimators": [100, 200]},
    {"vectorizer": vectorizers, "classifier": [SVC(kernel="linear")],
     "classifier__C": [0.1, 1]},
    {"vectorizer": [TfidfVectorizer(max_features=300)], "classifier": [CustomDecisionTree()],
     "classifier__max_depth": [8, 12]},
]

grid_search = GridSearchCV(pipeline, param_grid, scoring="f1", cv=5, n_jobs=-1)
grid_search.fit(X_train_text, y_train)

print(f"Meilleure combinaison : {grid_search.best_params_}")
print(f"Meilleur F1-score (validation croisée) : {grid_search.best_score_:.3f}")
```

```python
results = pd.DataFrame(grid_search.cv_results_)
results["classifier_name"] = results["param_classifier"].apply(lambda c: type(c).__name__)
results["vectorizer_name"] = results["param_vectorizer"].apply(lambda v: type(v).__name__)

# Meilleur score obtenu par chaque modele, toutes vectorisations/hyperparametres confondus
summary = (
    results.groupby("classifier_name")["mean_test_score"]
    .max()
    .sort_values(ascending=False)
    .rename("Meilleur F1-score (validation croisée)")
)
summary
```

```python
plt.figure(figsize=(8, 4.5))
colors = ["#C44E52" if name == grid_search.best_params_["classifier"].__class__.__name__ else "#4C72B0" for name in summary.index]
plt.barh(summary.index, summary.values, color=colors)
plt.xlabel("F1-score (validation croisée)")
plt.title("Meilleur F1-score obtenu par modèle")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

**Interprétation :** le meilleur F1-score en validation croisée est obtenu par la **Régression Logistique + TF-IDF** (F1 ≈ 0,738), suivie de près par le **SVM linéaire** (≈ 0,731), la **Random Forest** (≈ 0,728) et **XGBoost** (≈ 0,714). Le **Decision Tree** seul de scikit-learn est nettement plus faible (≈ 0,687), et notre **CustomDecisionTree** ferme la marche (≈ 0,431), pénalisé par sa limite volontaire à 300 mots. Les modèles linéaires et les modèles d'ensemble dominent donc largement les arbres de décision isolés sur ce texte à haute dimension. Le `TfidfVectorizer` l'emporte légèrement sur le `CountVectorizer` pour la plupart des modèles, ce qui confirme l'intérêt de pondérer les mots par leur importance plutôt que par leur simple fréquence (voir veille, question 8). Nous détaillons l'évaluation du meilleur modèle dans la section suivante.

### 8. Evaluation des modèles avec plusieurs métriques

Nous évaluons maintenant nos modèles sur le jeu de **validation** (des tweets jamais vus pendant l'entraînement ni le GridSearch), avec plusieurs métriques :
- **Accuracy** : proportion de tweets correctement classés.
- **Précision** : parmi les tweets prédits "catastrophe", combien le sont vraiment.
- **Recall** : parmi les vrais tweets "catastrophe", combien ont été détectés.
- **F1-score** : moyenne harmonique de la précision et du recall, la métrique que nous avons optimisée dans le GridSearch (`scoring="f1"`).

```python
from sklearn.base import clone
from sklearn.metrics import precision_score, recall_score

# On reprend, pour chaque famille de modele, sa meilleure combinaison trouvee par le GridSearch
best_per_model = results.loc[results.groupby("classifier_name")["mean_test_score"].idxmax()]

eval_rows = []
for _, row in best_per_model.iterrows():
    model = clone(pipeline).set_params(**row["params"])
    model.fit(X_train_text, y_train)
    preds = model.predict(X_val_text)
    eval_rows.append({
        "Modèle": row["classifier_name"],
        "Accuracy": accuracy_score(y_val, preds),
        "Précision": precision_score(y_val, preds),
        "Recall": recall_score(y_val, preds),
        "F1-score": f1_score(y_val, preds),
    })

eval_df = pd.DataFrame(eval_rows).sort_values("F1-score", ascending=False).reset_index(drop=True)
eval_df.round(3)
```

#### Matrice de confusion et rapport de classification du meilleur modèle

```python
from sklearn.metrics import classification_report, confusion_matrix

best_model = grid_search.best_estimator_
best_model_name = type(best_model.named_steps["classifier"]).__name__
best_preds = best_model.predict(X_val_text)

print(f"Rapport de classification - {best_model_name} :\n")
print(classification_report(y_val, best_preds, target_names=["Non-catastrophe", "Catastrophe"]))

cm = confusion_matrix(y_val, best_preds)

plt.figure(figsize=(5, 4.5))
plt.imshow(cm, cmap="Blues")
plt.title(f"Matrice de confusion - {best_model_name}")
plt.xlabel("Prédiction")
plt.ylabel("Réel")
plt.xticks([0, 1], ["Non-catastrophe", "Catastrophe"])
plt.yticks([0, 1], ["Non-catastrophe", "Catastrophe"])
for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha="center", va="center",
                  color="white" if cm[i, j] > cm.max() / 2 else "black")
plt.colorbar()
plt.tight_layout()
plt.show()
```

**Interprétation :** sur les 1523 tweets de validation, le modèle (Régression Logistique) classe correctement 775 des 869 tweets non-catastrophe et 471 des 654 tweets de catastrophe. Il se trompe dans deux sens : 94 tweets non-catastrophe sont prédits à tort comme catastrophe (faux positifs), et surtout 183 vrais tweets de catastrophe sont prédits comme non-catastrophe (faux négatifs). Ces derniers sont les plus problématiques pour notre problématique, car ce sont de vraies informations d'urgence qui passeraient inaperçues : le recall sur la classe "Catastrophe" (0,72) est d'ailleurs plus faible que sa précision (0,83), ce qui confirme que le modèle est un peu plus prudent (il rate quelques vraies catastrophes) que sur-alerte (peu de fausses alertes).

- Les modèles linéaires (Régression Logistique, SVM) et les modèles d'ensemble (Random Forest, XGBoost) se comportent nettement mieux que les arbres de décision seuls (sklearn ou maison) sur ce jeu de données textuel à haute dimension.
- Le `TfidfVectorizer` donne globalement de meilleurs résultats que le `CountVectorizer`, ce qui confirme l'intérêt de pondérer les mots rares mais discriminants plutôt que de compter simplement leurs occurrences.

**Modèle retenu : Régression Logistique + TF-IDF (3000 mots, unigrammes + bigrammes, C=1).** C'est elle qui obtient le meilleur F1-score (0.773) en validation, la métrique demandée par la problématique car elle équilibre précision et recall — deux erreurs coûteuses ici : rater un vrai tweet de catastrophe (faux négatif) ou noyer les secours sous de fausses alertes (faux positif). Elle est de plus simple, rapide à entraîner et facile à interpréter, ce qui en fait un choix pragmatique pour un premier outil de tri.

**Limites et pistes d'amélioration :**

- Des approches plus récentes (embeddings de mots, modèles de langage pré-entraînés type BERT) pourraient encore améliorer les performances, au prix d'une complexité et d'un coût de calcul plus élevés.
